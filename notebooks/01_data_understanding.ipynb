## 10. Calculate Total Units Sold by SKU

`groupby('sku_id')` splits the data by SKU and `.sum()` adds up the units sold.
This tells us which product sells the most overall.

```python
total_units_sold = sales_daily.groupby('sku_id')['units_sold'].sum()
print('Total units sold by SKU:')
print(total_units_sold)
```

## 11. Calculate Average Daily Units Sold by SKU

`.mean()` gives the average units sold per day for each SKU.
This is a simple, easy-to-understand baseline we can use later to compare against forecasts.

```python
avg_daily_units_sold = sales_daily.groupby('sku_id')['units_sold'].mean()
print('Average daily units sold by SKU:')
print(avg_daily_units_sold)
```

## 13. Display the Inventory Data

The inventory table tracks stock movement by day:
- `opening_stock` = stock at the start of the day
- `units_received` = units that arrived that day
- `units_sold` = units sold that day
- `closing_stock` = stock at the end of the day

The stock flow formula is:  `closing_stock = opening_stock + units_received - units_sold`

We first display the table, then run a read-only check to confirm the formula holds for every row.

```python
print('--- First 5 rows ---')
inventory_snapshots.head()

print('--- Last 5 rows ---')
inventory_snapshots.tail()
```

```python
# Read-only check: does closing_stock always equal opening + received - sold?
calculated_closing = (
    inventory_snapshots['opening_stock']
    + inventory_snapshots['units_received']
    - inventory_snapshots['units_sold']
)

print('Formula holds for every row:', (calculated_closing == inventory_snapshots['closing_stock']).all())
```
